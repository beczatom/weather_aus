## Predstavenie príznakov
Dáta obsahujú 22 príznakov a jednu vysvetľovanú premennú
- `Date` - dátum merania
- `Location` - názov miesta merania
- `MinTemp` - minimálna teplota za celý deň (v °C)
- `MaxTemp` - maximálna teplota za celý deň (v °C)
- `Rainfall` - úhrn zrážok (v mm)
- `Evaporation` - množstvo vyparenej vody (v mm)
- `Sunshine` - počet hodín jasného slnečného svitu
- `WindGustDir` - smer najsilnejšieho nárazu vetra (v smere podľa kompasu (sever, juhovýchod, ...))
- `WindGustSpeed` - rýchlosť najsilnejšieho vetra (v km/h)
- `WindDir9am` - smer vetra o 9. hodine (v smere podľa kompasu (sever, juhovýchod, ...))
- `WindDir3pm` - smer vetra o 15. hodine (v smere podľa kompasu (sever, juhovýchod, ...))
- `WindSpeed9am` - priemerná rýchlosť vetra počas 10 minút pred 9. hodinou (v km/h)
- `WindSpeed3pm` - priemerná rýchlosť vetra počas 10 minút pred 15. hodinou (v km/h)
- `Humidity9am` - vlhkosť o 9. hodine (v %)
- `Humidity3pm` - vlhkosť o 15. hodine (v %)
- `Pressure9am` - atmosférický tlak o 9. hodine (v hPa)
- `Pressure3pm` - atmosférický tlak o 15. hodine (v hPa)
- `Cloud9am` - oblačnosť o 9. hodine (v "okta" (osminách), vyjadruje počet osmín oblohy pokrytého oblakmi (0 znamená žiadnu oblačnosť a 8 vyjadruje úplne zamračenú oblohu))
- `Cloud3pm` - oblačnosť o 15. hodine (v "okta" (osminách), vyjadruje počet osmín oblohy pokrytého oblakmi (0 znamená žiadnu oblačnosť a 8 vyjadruje úplne zamračenú oblohu))
- `Temp9am` - teplota o 9. hodine (v °C)
- `Temp3pm` - teplota o 15. hodine (v °C)
- `RainToday` - vyjadruje, či pršalo v deň merania (yes/no)
- `RainTomorrow` - *vysvetľovaná premenná*, vyjadruje, či pršalo v deň, ktorý bezprostredne nasleduje po dni merania (yes/no)


## Predspracovanie

Odstránenie dát s chýbajúcou vysvetľovanou premennou

```python
df = df[~(df['RainTomorrow'].isnull())]
```

### Chýbajúce hodnoty

Pozriem sa na to, ktoré príznaky majú koľko percent nevyplnených polí.

```python
null_val = []
for col in df.columns:
    null_val.append(df[col].isnull().sum() / df.shape[0] * 100)

fig = plt.figure(figsize = (10,6))
ax = fig.add_subplot()

ax.set_facecolor('#EBFFEB')

ax.barh(df.columns, null_val, facecolor='#6CA6C1')

ax.set_title('Null values distribution')
ax.set_xlabel('Null values [%]')
ax.set_ylabel('Feature name')
plt.show()
```

Môžeme vidieť, že väčšina príznakov je relatívne vyplnená. Vytŕčajú hlavne `Sunshine`, `Evaporation` a hodnoty oblačnosti.

### Doplnenie datasetu
Príznak `Location` nie je v najvhodnejšom stave pre strojové učenie, napriek tomu, že je pre predpoveď určite veľmi dôležitý, pridám preto jednotlivé súradnice.

```python
df['Location'].unique()
```

Načítam dáta z ďalšieho súboru, ktorý obsahuje názov miesta a jeho zemepisnú šírku a dĺžku.

```python
coords = pd.read_csv('au.csv')
coords = coords[['city', 'lat', 'lng']]
coords.rename(columns={'city': 'Location'}, inplace=True)
coords
```

"Prilepím" ich k dátam o počasí.

```python
df = df.join(coords.set_index('Location'), on='Location')
df.info()
```

Príznak `Date` je uložený v dátovom type *string*, to tiež nie je veľmi vhodné, preto ho rozdelím na deň, mesiac a rok.

```python
df['Date'] = pd.to_datetime(df['Date'])

df['day'] = df['Date'].dt.day
df['month'] = df['Date'].dt.month
df['year'] = df['Date'].dt.year

df.drop(columns=['Date'], inplace=True)
df
```

Následne tu máme kategorické príznaky `WindGustDir`, `WindDir9am` a `WindDir3pm`, ktoré by po one-hot encodingu podľa môjho názoru narobili viac škody ako úžitku (zvyšovanie dimenzie), preto ich odtránim.

```python
df = df.drop(['WindGustDir', 'WindDir9am', 'WindDir3pm'], axis=1)
df
```

O vhodnosti reprezentovať skutočnosť, či prší, reťazcami *yes* a *no* sa snáď ani netreba baviť, skonvertujem ich na typ *bool*.

```python
df['RainToday'] = df['RainToday'] == 'Yes'
df['RainTomorrow'] = df['RainTomorrow'] == 'Yes'
df.info()
```

Teraz sa pozriem bližšie na priemernú teplotu v jednotlivých meracích staniciach.

```python
# group by locatiom
loc_group = df.groupby('Location')
loc_temps = []
lats = []
lngs = []

# for each location
for name, group in loc_group:
    # get coordinates
    lats.append(group['lat'].iloc[0])
    lngs.append(group['lng'].iloc[0])

    # append mean ( mean min temp + mean max temp )
    loc_temps.append(group['MinTemp'].mean())
    loc_temps[np.size(loc_temps) - 1] += group['MaxTemp'].mean()
    loc_temps[np.size(loc_temps) - 1] /= 2

# df for easier visualization
df_map = pd.DataFrame({
    'lat' : lats,
    'lng' : lngs,
    'temperature' : loc_temps
})

# visualize map
fig = px.scatter_mapbox(
    df_map,

    # marker pos coordinates
    lat = 'lat',
    lon = 'lng',

    # marker size and color
    color = 'temperature',
    size = 'temperature',
    title = 'Average year temperature',

    # not exactly center of Australia, rather center of our points
    center = {'lat': -24, 'lon': 140},
    width = 800,
    height = 700,
    zoom = 2.75,
    mapbox_style = 'open-street-map'
)
fig.show()
```

Pozrime sa teraz na rozdelenie úhrnu zrážok medzi mesiace pre pár väčších miest.

```python
# for displaying rainfall by month by city
groups = df.groupby(['Location', 'month'])

rain = {}
for city in df['Location'].unique():
    rain[city] = {}

# for each city we need 12 averages
for name, group in groups:
    rain[name[0]][name[1]] = group['Rainfall'].mean()

fig = plt.figure(figsize=(18,10), facecolor='#EBFFEB')
gs = GridSpec(2, 2, figure=fig)

def rain_bar(ax, city):
    ax.bar(x = np.arange(1, 13, 1), height = rain[city].values(), facecolor='#2F3061')
    ax.set_facecolor('#EBFFEB')
    ax.set_title(city)
    ax.set_xticks(np.arange(1, 13, 1))
    ax.set_xlabel('Month')
    ax.set_ylabel('Average rainfall per day [mm]')
    ax.set_ylim(0, 16)

# just some major cities
rain_bar(fig.add_subplot(gs[0, 0]), 'Sydney')
rain_bar(fig.add_subplot(gs[0, 1]), 'Perth')
rain_bar(fig.add_subplot(gs[1, 0]), 'Brisbane')
rain_bar(fig.add_subplot(gs[1, 1]), 'Darwin')
plt.show()
```

Vlhkosť je iste dôležitý činiteľ pre možnosť dažďa, poďme sa pozrieť na jeho priemerné hodnoty v závislosti na oblasti.

```python
# shows australian map divided with lats and lngs
# for each cell (created by lats and lngs) sets its respective color
def aus_map_values_plot(lats, lngs, values, min_val, max_val, title, bar_title):
    # normalization for color usage
    norm = colors.Normalize(vmin=min_val, vmax=max_val)
    cmap = pyplot.get_cmap('YlGn')

    # for each rectangle in the grid
    rects = []
    for i in range(1, len(lats)):
        for j in range(1, len(lngs)):
            # if not present value -> transparent rectangle
            color = f'rgba(0,0,0,0)'

            # values is set
            if not np.isnan(values[i - 1][j - 1]):
                # set color to a rectangle
                rgba = cmap(norm(values[i - 1][j - 1]))
                color = f'rgba({int(rgba[0] * 255)}, {int(rgba[1] * 255)}, {int(rgba[2] * 255)}, {rgba[3] * 0.8})'

            # append to other rectangles with correct values
            rects.append(
                dict(
                    type='rect',

                    # longitude aka x-axis
                    x0 = lngs[j - 1],
                    x1 = lngs[j],

                    # latitude aka y-axis
                    y0 = lats[i - 1],
                    y1 = lats[i],
                    xref='x',
                    yref='y',

                    # correct color
                    fillcolor=color,
                    line=dict(width=0.2)
                )
            )

    fig = go.Figure()

    # sets "background image" -> Australia
    fig.update_layout(
        title=title,
        width=800,
        height=520,
        geo=dict(
            # map type
            projection_type="equirectangular",

            # map style
            showcountries=True,
            landcolor='#999999',
            oceancolor='#BBBBDD',

            # snap to correct position
            lonaxis=dict(range=[min(lngs), max(lngs) - 0.1]),
            lataxis=dict(range=[min(lats), max(lats) - 0.1]),
        ),
        shapes = rects,
    )

    # empty Scattergeo, only for colorscale
    fig.add_trace(
        go.Scattergeo(
            lon = [None],
            lat = [None],
            mode='markers',
            marker=dict(
                # setting the colorscale
                cmin = min_val,
                cmax = max_val,
                colorscale = 'ylgn',
                showscale = True,
                colorbar = dict(
                    title=bar_title,
                )
            ),
            hoverinfo='none',
        )
    )
    fig.show()
```

```python
# defining the grid
lats = np.linspace(df['lat'].min(), df['lat'].max() + 0.1, 8)
lngs = np.linspace(df['lng'].min(), df['lng'].max() + 0.1, 10)

hum = []
min_hum = 1000
max_hum = 0

# for each cell in grid
for i in range(1, len(lats)):
    hum.append([])
    for j in range(1, len(lngs)):
        # gets only data with location in grid
        mask = df['lat'].between(lats[i - 1], lats[i], inclusive='left') * df['lng'].between(lngs[j - 1], lngs[j], inclusive='left')

        # gets average from this data
        hum[i - 1].append(df[mask]['Humidity3pm'].mean())

    if max(hum[-1]) > max_hum:
        max_hum = max(hum[-1])
    if min(hum[-1]) < min_hum:
        min_hum = min(hum[-1])

aus_map_values_plot(lats, lngs, hum, min_hum, max_hum, 'Average humidity levels in areas', 'Humidity [%]')
```

V strede Austrálie vidíme oblasť s nízkou vlhkosťou, neprekvapivo, keďže tam je púšť. Čím bližšie sme k pobrežiu, tým viac stúpa vlhkosť, čo je tiež očakávaný jav.

# Model
## Doplnenie príznakov
Pre jednotlivé dvojice príznakov, pre ktoré to dáva zmysel, mi príde ako dobrý nápad, doplniť pár jednoduchých lineérnych transformácii príznakov. Konkrétne rôzne zmeny počas dňa. Je podľa mňa pravdepodobnejšie, že bude pršať na druhý deň, ak sa počasie už predtým deň menilo.

```python
df['WindSpeedDelta'] = df['WindSpeed9am'] - df['WindSpeed3pm']
df['HumidityDelta'] = df['Humidity9am'] - df['Humidity3pm']
df['PressureDelta'] = df['Pressure9am'] - df['Pressure3pm']
df['CloudDelta'] = df['Cloud9am'] - df['Cloud3pm']
df['TempDelta'] = df['Temp9am'] - df['Temp3pm']
```

## Rozdelenie dát
Dáta rozdelím na trénovaciu (na ktorej model natrénujem), na validačnú (na ktorej budem model hodnotiť) a na testovaciu množinu (na ktorej zmeriam odhadovanú presnosť modelu na dátach, ktoré ešte model nevidel).

```python
rd_seed = 28
Xdata = df.drop(['RainTomorrow', 'Location', 'RainTomorrow'], axis=1)
Xdata = pd.get_dummies(Xdata)
ydata = df['RainTomorrow']

Xdata, Xtest, ydata, ytest = train_test_split(Xdata, ydata, test_size=0.3, random_state=rd_seed)
Xtrain, Xval, ytrain, yval = train_test_split(Xdata, ydata, test_size=0.3, random_state=rd_seed)

print('Train shape:\t\t', Xtrain.shape)
print('Validation shape:\t', Xval.shape)
print('Test shape:\t\t\t', Xtest.shape)
```

## Výber modelu
Pre výber modelu na klasifikáciu máme viacero možností. Najprv si vyberiem *rozhodovací strom*, pretože nemám príliš veľa príznakov (nebude mu prekážať vysoká dimenzia) a je relatívne jednoduchý na pochopenie. Rozhodovací strom delí postupne trénovaciu množinu na menšie pomocou takých hodnôt príznakov, ktoré ju najlepšie rozdelia (na meranie ako je delenie "dobré" budem používať entropiu). Cieľom teda bude znížiť entropiu na čo najmenšiu hodnotu.

### Výber hyperparametrov
Na trénovanie stromu budem používať hyperparametre:
- *max_depth* - ako maximálnu hĺbku stromu. Stromu chceme zabrániť aby sa príliš sústredil na nejaké nepodstatné odlišnosti v trénovacej množine, pretože tie sa nemusia prejaviť v reálnych dátach (preučenie). Vyskúšam každú hodnotu od 1 do 10, očakávam, že niekde tu nájdem maximálnu presnosť a po tej hodnote bude už presnosť na validačnej množine klesať.
- *min_samples_split* - tak ako *max_depth* aj tento hyperparameter zabraňuje preučeniu, vyjadruje koľko musí mať aspoň množina dátových bodov, aby sa ju mohli rozdeliť. Vyskúšam hodnoty od 2 po 200, po skokoch 40, očakávam, že s hodnotami nad 100, už bude model značne nepresný.


```python
hyper_params = {
    'max_depth' : range(1, 11),
    'min_samples_split' : range(2, 200, 40),
}
param_combs = ParameterGrid(hyper_params)
```

### Trénovanie
Pre rôzne kombinácie hyperparametrov teraz model natrénujem.

```python
val_accs = {}
train_accs = {}

# for each param
for param in param_combs:
    tree = DecisionTreeClassifier(**param, random_state=rd_seed)
    tree.fit(Xtrain, ytrain)

    # append accuracy
    val_accs[tuple(param.values())] = tree.score(Xval, yval)
    train_accs[tuple(param.values())] = tree.score(Xtrain, ytrain)
```

Zopár funkcií pre krajšiu vizualizáciu výsledkov trénovania.

```python
# get best params and accuracy from above generated dictionary
def get_best_params_acc(accs):
    best_params = max(accs, key=accs.get)
    return best_params, accs[best_params]
```

```python
# plots accuracy on validation and train set
def plot_accuracy(val_accs, train_accs, ax):
    best_params, best_acc = get_best_params_acc(val_accs)

    ax.plot(range(0, len(val_accs)), val_accs.values(), '-o', color='#30305F', label='Validation Accuracy')
    ax.plot(range(0, len(train_accs)), train_accs.values(), '-o', color='#FFDA1F', label='Train Accuracy')

    # showing the maximum
    ax.plot([0, len(val_accs) - 1], [best_acc, best_acc], '--g')
    ax.annotate(f'Best accuracy: {best_acc:.3f}', (0, best_acc), xytext=[1, best_acc * 1.002], size=14)

    ax.set_xlabel('Hyperparameter combination index')
    ax.set_ylabel('Accuracy')
    ax.legend(loc='best')
    ax.set_title('Accuracy of model')

    return best_params, best_acc
```

```python
# plots 2D scatter of hyperparams and colors it with their accuracy on validation set
def scatter_accuracy(val_accs, ax, xlabel, ylabel):
    best_params, best_acc = get_best_params_acc(val_accs)

    # get x and y axis
    x_ax = [param[0] for param in val_accs.keys()]
    y_ax = [param[1] for param in val_accs.keys()]

    # colors
    norm = colors.Normalize(vmin=min(val_accs.values()), vmax=max(val_accs.values()))
    cmap = pyplot.get_cmap('YlGn')

    scatter = ax.scatter(x=x_ax, y=y_ax, c=val_accs.values(), norm=norm, cmap=cmap, s=70, edgecolors='#3E3E7A')

    # showing the best combination
    ax.annotate(f'Best accuracy: {best_acc:.3f}', (best_params[0], best_params[1]), xytext=[best_params[0] - best_params[0] * 0.12, best_params[1] + best_params[1] * 2], size=10)

    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)

    # showing color legend
    plt.colorbar(scatter, label='Accuracy')
```

### Závislosť presnosti na hodnote hyperparametrov

```python
fig = plt.figure(figsize=(10, 10), facecolor='#EBFFEB')
gs = GridSpec(2, 1, figure=fig)

# plot
ax1 = fig.add_subplot(gs[0, :])
ax1.set_facecolor('#EBFFEB')
plot_accuracy(val_accs, train_accs, ax1)

# scatter
ax2 = fig.add_subplot(gs[1, 0])
ax2.set_facecolor('#EBFFEB')
scatter_accuracy(val_accs, ax2, 'max_depth', 'min_samples_split')

best_params, best_acc = get_best_params_acc(val_accs)
print('Best params:\tmax_depth:\t', best_params[0], '\tmin_samples_split:\t', best_params[1])
print(f'Best accuracy:\t{best_acc:.3f}')
plt.show()
```

Vidíme, ako rapídne skokovo sa zvyšuje presnosť modelu pre pribúdajúcu hĺbku stromu. No od hĺbky 7 dochádza k zastaveniu rastu presnosti na validačnej množine. To signalizuje preučenie modelu. Teda berieme ako najlepší strom práve tento.

```python
best_tree = DecisionTreeClassifier(criterion='entropy', max_depth=best_params[0], min_samples_split=best_params[1], random_state=rd_seed)
best_tree.fit(Xtrain, ytrain)
```

### Evaluácia
Pozrieme sa presnejšie ako náš model v skutočnosti je presný.

```python
# displays confusion matrix and relative confusion matrix on Xset
def conf_matrices(model, Xset, yset):
    # get the prediction
    ypred = model.predict(Xset)

    # get confusion matrix values
    conf = confusion_matrix(y_true=yset, y_pred=ypred)

    # styles
    fig = plt.figure(figsize=(12, 8), facecolor='#EBFFEB')
    gs = GridSpec(1, 2, figure=fig)
    ax1 = fig.add_subplot(gs[0, 0])
    ax1.grid(visible=False)
    ax1.set_title('Confusion matrix')
    ax2 = fig.add_subplot(gs[0, 1])
    ax2.grid(visible=False)
    ax2.set_title('Normalized confusion matrix')

    # display confusion matrix
    matrix_disp = ConfusionMatrixDisplay(confusion_matrix=conf, display_labels=['0', '1'])
    matrix_disp.plot(ax=ax1, cmap='YlGn')

    # display normalized confusion matrix by true values
    matrix_disp_norm = ConfusionMatrixDisplay.from_predictions(y_true=yset, y_pred=ypred, display_labels=['0', '1'], normalize='true', ax=ax2, cmap='YlGn')
```

Pekne to bude vidno aj na ROC krivke a s ním spojenou AUC hodnotou.

```python
# plots ROC curve on Xset
def plot_roc(model, Xset, yset):

    # get the probabilities
    ypred_probs = model.predict_proba(Xset)[:, -1]

    # compute roc curve
    fpr, tpr, treshold = roc_curve(yset, ypred_probs)

    fig = plt.figure()
    ax = fig.add_subplot()

    # roc curve itself
    ax.plot(fpr, tpr, label='ROC curve', color='#30305F')

    # styles
    ax.set_facecolor('#EBFFEB')
    ax.set_xlabel('False Positive Rate')
    ax.set_ylabel('True Positive Rate')
    ax.title.set_text('ROC Curve')
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.0])
    ax.set_xticks(np.arange(0, 1.1, 0.2))
    ax.set_yticks(np.arange(0, 1.1, 0.2))

    # ideal random prediction line
    ax.plot([0, 1], [0, 1], 'r--')
    ax.legend(loc='lower right')
    plt.show()

    # AUC value
    auc_model = auc(fpr, tpr)
    print('AUC value:', round(auc_model, 3))

    # predict without probabilities for F1 score
    ypred = model.predict(Xset)

    # F1 score
    F1_model = f1_score(yset, ypred)
    print('F1 score:', round(F1_model, 3))
```

Obe tieto hodnoty sú závislé na tzv. tresholde, teda na prahom pravdepodobnosti jedničky v istom liste stromu, nad ktorým už predpovedáme jedničku. Implicitne predpovedáme 1 ak je počet jedničiek v liste väčší alebo rovný 1/2.

## Logistická regresia
Pre vyššie opísané dôvody, som sa rozhodol použiť aj model logistickej regresie. Logistická regresia je klasifikačný model, ktorý predpokladá lineárny vzťah medzi príznakmi a vysvetľovanou premennou. Túto rovnicu potom dosadí do sigmoidy, odkiaľ máme hodnoty 0-1, vhodné pre predpovedanie pravdepodobnosti, jednotlivé pravdepodobnosti potom maximalizuje metódou najväčšej vierohodnosti. Ja si zvolím jediný hyperparameter a to je vyrovnanosť váh tried, teda aby boli presnosti pre daždivé a suché dni čo najvyrovnanejšie a samozrejme čo najvyššie také.

Dáta najprv normalizujem. Pri strome to nebolo nutné, keďže on len porovnáva hodnoty a normalizáciou usporiadanie nezmením. Takisto by to nebolo potrebné ani tu, logistická regresia si prípadné zmeny hodnôt ajtak premietne do zmeny koeficientov. Štandardizáciu potrebujem ale pretože neskôr budem porovnávať dôležitosť príznakov pomocou veľkosti koeficientov. Ak by hodnoty nechal ako sú, tak by to bol "nerovný súboj" pre príznaky s tradične vyššou hodnotou (napr. *Pressure* v tisíckach) a príznaky s nižšou hodnotou (napr. *RainToday* s maximálnou hodnotou 1).

```python
# scale the train set
scaler = MinMaxScaler()
Xtrain_norm = scaler.fit_transform(Xtrain)
Xval_norm = scaler.transform(Xval)

# logistic regression "training"
lreg = LogisticRegression(class_weight='balanced', solver='liblinear')
lreg.fit(Xtrain_norm, ytrain)
print(f'Accuracy: {lreg.score(Xval_norm, yval):.3f}')
```

Pozrieme sa ešte na úspešnosť modelov v jednotlivých oblastiach Austrálie, je možné očakávať lepšiu presnosť stromu pre suchšie oblasti.

```python
# gets prediction accuracy for cell in grid generated by lats and lngs
# !note i use Xset and Xset norm, because the selection by lats and lngs in normalized set is useless
def get_acc_in_areas(model, Xset, Xset_norm, yset, lats, lngs):
    acc = []
    min_acc = 100
    max_acc = 0

    # for each cell
    for i in range(1, len(lats)):
        acc.append([])
        for j in range(1, len(lngs)):
            # data that are in area
            # !note Xset is used only for selection of data
            mask = Xset['lat'].between(lats[i - 1], lats[i], inclusive='left') * Xset['lng'].between(lngs[j - 1], lngs[j], inclusive='left')

            # empty area
            if Xset[mask].shape[0] == 0:
                acc[i - 1].append(np.nan)
                continue

            # append score in percent
            # !note the model is actually validated on normalized set
            acc[i - 1].append(model.score(Xset_norm[mask], yset[mask]) * 100)

        if max(acc[-1]) > max_acc:
            max_acc = max(acc[-1])
        if min(acc[-1]) < min_acc:
            min_acc = min(acc[-1])

    return acc, min_acc, max_acc
```
