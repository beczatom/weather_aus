```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('weatherAUS.csv')
df
```

```python
df.info()
```

```python
df['Location'].unique()
```

```python
coords = pd.read_csv('au.csv')
```

```python
coords = coords[['city', 'lat', 'lng']]
```

```python
coords.rename(columns={'city': 'Location'}, inplace=True)
```

```python
coords
```

```python
df = df.join( coords.set_index('Location'), on='Location')
```

```python
df.info()
```

```python
df[df['lat'].isnull()]['Location'].unique()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df['day'] = df['Date'].dt.day
df['month'] = df['Date'].dt.month
df['year'] = df['Date'].dt.year
df.drop(columns=['Date'], inplace=True)
df
```

```python
df['WindGustDir'].value_counts()
```

```python
df['WindGustDir'] = df['WindGustDir'].astype('category')
df['WindDir9am'] = df['WindDir9am'].astype('category')
df['WindDir3pm'] = df['WindDir3pm'].astype('category')
df
```

```python
df['RainToday'] = df['RainToday'] == 'Yes'
df['RainTomorrow'] = df['RainTomorrow'] == 'Yes'
```

```python
df.info()
df.duplicated().sum()
```

```python
for column in df.select_dtypes(exclude = ['object', 'category']).columns:
    df[column] = df[column].fillna(df[column].mean())
```

```python
df
```

```python
fig = plt.figure(figsize=[8,8])
plt.boxplot([df['MinTemp'], df['MaxTemp']], tick_labels=['min', 'max'])
plt.show()
```

```python
loc_group = df.groupby('Location')
loc_temps = []
lats = []
lngs =[]
for name, group in loc_group:
    lats.append(group['lat'].iloc[0])
    lngs.append(group['lng'].iloc[0])
    loc_temps.append(group['MinTemp'].mean())
    loc_temps[np.size(loc_temps) - 1] += group['MaxTemp'].mean()
    loc_temps[np.size(loc_temps) - 1] /= 2

print(loc_temps)
```

```python
import plotly.express as px
fig = px.scatter_geo(
    lat= lats,
    lon= lngs,
    color= loc_temps,
    size= loc_temps,
    title='Average year temperature',
    # discrimination of australia, every other continent can be used as scope
    # scope='australia',
    scope='asia',
    center= {'lat': -25.2744, 'lon': 133.751},
    width=800,
)
fig.show()
```
