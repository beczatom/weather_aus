```python
def aus_map_values_plot(lats, lngs, values, min_val, max_val, title, bar_title):
    norm = colors.Normalize(vmin=min_val, vmax=max_val)
    cmap = pyplot.get_cmap('YlGn')

    rects = []
    for i in range(1, len(lats)):
        for j in range(1, len(lngs)):
            color = f'rgba(0,0,0,0)'
            if not np.isnan(values[i - 1][j - 1]):
                rgba = cmap(norm(values[i - 1][j - 1]))
                color = f'rgba({int(rgba[0] * 255)}, {int(rgba[1] * 255)}, {int(rgba[2] * 255)}, {rgba[3] * 0.8})'

            rects.append(
                dict(
                    type='rect',
                    x0 = lngs[j - 1],
                    x1 = lngs[j],
                    y0 = lats[i - 1],
                    y1 = lats[i],
                    xref='x',
                    yref='y',
                    fillcolor=color,
                    line=dict(width=0.2)
                )
            )

    fig = go.Figure()
    fig.update_layout(
        title=title,
        width=800,
        height=520,
        geo=dict(
            projection_type="equirectangular",
            showcountries=True,
            landcolor='#999999',
            oceancolor='#BBBBDD',
            lonaxis=dict(range=[min(lngs), max(lngs) - 0.1]),
            lataxis=dict(range=[min(lats), max(lats) - 0.1]),
        ),
        shapes = rects,
    )

    fig.add_trace(
        go.Scattergeo(
            lon = [None],
            lat = [None],
            mode='markers',
            marker=dict(
                cmin = min_val,
                cmax = max_val,
                colorscale = 'ylgn',
                showscale = True,
                colorbar = dict(
                    title=bar_title,
                )
            ),
            hoverinfo='none',
        )
    )
    fig.show()
```

# Model

```python
df['WindSpeedDelta'] = df['WindSpeed9am'] - df['WindSpeed3pm']
df['HumidityDelta'] = df['Humidity9am'] - df['Humidity3pm']
df['PressureDelta'] = df['Pressure9am'] - df['Pressure3pm']
df['CloudDelta'] = df['Cloud9am'] - df['Cloud3pm']
df['TempDelta'] = df['Temp9am'] - df['Temp3pm']
```

```python
rd_seed = 28
Xdata = df.drop(['RainTomorrow', 'Location', 'RainTomorrow'], axis=1)
Xdata = pd.get_dummies(Xdata)
ydata = df['RainTomorrow']
Xdata, Xtest, ydata, ytest = train_test_split(Xdata, ydata, test_size=0.3, random_state=rd_seed)
Xtrain, Xval, ytrain, yval = train_test_split(Xdata, ydata, test_size=0.3, random_state=rd_seed)
Xtrain.shape, Xval.shape, Xtest.shape
```

```python
from sklearn.model_selection import ParameterGrid

hyper_params = {
    'max_depth' : range(1, 10),
    'min_samples_split' : range(2, 100, 20),
}
param_combs = ParameterGrid(hyper_params)
```

```python
val_accs = {}
train_accs = {}
for param in param_combs:
    tree = DecisionTreeClassifier(**param, random_state=rd_seed)
    tree.fit(Xtrain, ytrain)
    val_accs[tuple(param.values())] = tree.score(Xval, yval)
    train_accs[tuple(param.values())] = tree.score(Xtrain, ytrain)
```

```python
def get_best_params_acc(accs):
    best_params = max(accs, key=accs.get)
    return best_params, accs[best_params]
```

```python
def plot_accuracy(val_accs, train_accs, ax):
    best_params, best_acc = get_best_params_acc(val_accs)

    ax.plot(range(0, len(val_accs)), val_accs.values(), '-o', color='#30305F', label='Validation Accuracy')
    ax.plot(range(0, len(train_accs)), train_accs.values(), '-o', color='#FFDA1F', label='Train Accuracy')
    ax.set_xlabel('Hyperparameter combination index')
    ax.set_ylabel('Accuracy')
    ax.plot([0, len(val_accs) - 1], [best_acc, best_acc], '--g')
    ax.annotate(f'Best accuracy: {best_acc:.3f}', (0, best_acc), xytext=[1, best_acc * 1.002], size=14)
    ax.legend(loc='best')
    ax.set_title('Accuracy of model')

    return best_params, best_acc
```

```python
def scatter_accuracy(val_accs, ax, xlabel, ylabel):
    best_params, best_acc = get_best_params_acc(val_accs)
    print(best_params, best_acc)
    x_ax = [param[0] for param in val_accs.keys()]
    y_ax = [param[1] for param in val_accs.keys()]
    norm = colors.Normalize(vmin=min(val_accs.values()), vmax=max(val_accs.values()))
    cmap = pyplot.get_cmap('YlGn')
    scatter = ax.scatter(x=x_ax, y=y_ax, c=val_accs.values(), norm=norm, cmap=cmap, s=70, edgecolors='#3E3E7A')
    ax.annotate(f'Best accuracy: {best_acc:.3f}', (best_params[0], best_params[1]), xytext=[best_params[0] - best_params[0] * 0.12, best_params[1] + best_params[1] * 2], size=10)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    plt.colorbar(scatter, label='Accuracy')
```

```python
fig = plt.figure(figsize=(10, 10), facecolor='#EBFFEB')
gs = GridSpec(2, 1, figure=fig)
ax1 = fig.add_subplot(gs[0, :])
ax1.set_facecolor('#EBFFEB')
plot_accuracy(val_accs, train_accs, ax1)
ax2 = fig.add_subplot(gs[1, 0])
ax2.set_facecolor('#EBFFEB')
scatter_accuracy(val_accs, ax2, 'max_depth', 'min_samples_split')
best_params, best_acc = get_best_params_acc(val_accs)
plt.show()
```

```python
best_tree = DecisionTreeClassifier(criterion='entropy', max_depth=best_params[0], min_samples_split=best_params[1], random_state=rd_seed)
best_tree.fit(Xtrain, ytrain)
```

```python
def conf_matrices(model, Xset, yset):
    ypred = model.predict(Xset)
    conf = confusion_matrix(y_true=yset, y_pred=ypred)

    fig = plt.figure(figsize=(12, 8), facecolor='#EBFFEB')
    gs = GridSpec(1, 2, figure=fig)
    ax1 = fig.add_subplot(gs[0, 0])
    ax1.grid(visible=False)
    ax1.set_title('Confusion matrix')
    ax2 = fig.add_subplot(gs[0, 1])
    ax2.grid(visible=False)
    ax2.set_title('Normalized confusion matrix')
    
    matrix_disp = ConfusionMatrixDisplay(confusion_matrix=conf, display_labels=['0', '1'])
    matrix_disp.plot(ax=ax1, cmap='YlGn')
    matrix_disp_norm = ConfusionMatrixDisplay.from_predictions(y_true=yset, y_pred=ypred, display_labels=['0', '1'], normalize='true', ax=ax2, cmap='YlGn')
```

```python
conf_matrices(best_tree, Xval, yval)
```

```python
def plot_roc(model, Xset, yset):

    ypred_probs = model.predict_proba(Xset)[:, -1]
    fpr, tpr, treshold = roc_curve(yset, ypred_probs)

    fig = plt.figure()
    ax = fig.add_subplot()
    ax.plot(fpr, tpr, label='ROC curve', color='#30305F')
    ax.set_facecolor('#EBFFEB')
    ax.set_xlabel('False Positive Rate')
    ax.set_ylabel('True Positive Rate')
    ax.title.set_text('ROC Curve')
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.0])
    ax.set_xticks(np.arange(0, 1.1, 0.2))
    ax.set_yticks(np.arange(0, 1.1, 0.2))
    ax.plot([0, 1], [0, 1], 'r--')
    ax.legend(loc='lower right')
    plt.show()

    # AUC value
    auc_model = auc(fpr, tpr)
    print('AUC value:', round(auc_model, 3))

    ypred = model.predict(Xset)
    # F1 score
    F1_model = f1_score(yset, ypred)
    print('F1 score:', round(F1_model, 3))
```

```python
def get_acc_in_areas(model, Xset, Xset_norm, yset, lats, lngs):
    acc = []
    min_acc = 100
    max_acc = 0
    for i in range(1, len(lats)):
        acc.append([])
        for j in range(1, len(lngs)):
            mask = Xset['lat'].between(lats[i - 1], lats[i], inclusive='left') * Xset['lng'].between(lngs[j - 1], lngs[j], inclusive='left')
            if Xset[mask].shape[0] == 0:
                acc[i - 1].append(np.nan)
                continue
            acc[i - 1].append(model.score(Xset_norm[mask], yset[mask]) * 100)
        if max(acc[-1]) > max_acc:
            max_acc = max(acc[-1])
        if min(acc[-1]) < min_acc:
            min_acc = min(acc[-1])
    return acc, min_acc, max_acc
```

```python
lats = np.linspace(Xtest['lat'].min(), Xtest['lat'].max() + 0.1, 8)
lngs = np.linspace(Xtest['lng'].min(), Xtest['lng'].max() + 0.1, 10)

acc_tree, min_acc_tree, max_acc_tree = get_acc_in_areas(best_tree, Xval, Xval, yval, lats, lngs)
acc_lreg, min_acc_lreg, max_acc_lreg = get_acc_in_areas(lreg, Xval, Xval_norm, yval, lats, lngs)
min_acc = min([min_acc_tree, min_acc_lreg])
max_acc = max([max_acc_tree, max_acc_lreg])
aus_map_values_plot(lats, lngs, acc_tree, min_acc, max_acc, 'Average accuracy in areas', 'Accuracy [%]')
aus_map_values_plot(lats, lngs, acc_lreg, min_acc, max_acc, 'Average accuracy in areas', 'Accuracy [%]')
```
